```python
import sys
from pathlib import Path

# this allows the importing of modules from ../modules
root_path = Path.cwd().parent
if str(root_path) not in sys.path:
    sys.path.append(str(root_path))

from modules import ImagenetLoader

loader = ImagenetLoader()
```

```python
model_50 = loader.load_radimagenet_resnet50(
    "../trained_models/radimagenet50_vqa_baseline1/radimagenet50_vqa_baseline1.pth",
    load_type="load"
    )

model_18 = loader.load_radimagenet_resnet18(
    "../trained_models/radimagenet18_vqa_baseline/radimagenet18_vqa_baseline.pth",
    load_type="load"
    )
```

```python
from modules import datasetPrepper, ModelEvaluator

data = datasetPrepper(
    "../data/labels.csv",
    image_dir="../data/test_images"
    ).prepare()

evaluator = ModelEvaluator(
    data_loader=data.val_loader,
    class_names=data.class_names
)

evaluator.evaluate_many(
    {"ResNet50_Baseline" : model_50,"ResNet18_Baseline" : model_18}
    )
```
